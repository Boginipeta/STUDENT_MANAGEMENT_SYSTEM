```python
# Student Record Management System

students = []   # list of dictionaries

 def add_student():
    sid = input("Enter Student ID: ")
    name = input("Enter Student Name: ")
    city = input("Enter City: ")
    student = {"id": sid, "name": name, "city": city}
    students.append(student)
    print("✅ Student added successfully!")

def view_students():
    if not students:
        print("No records found.")
    else:
        for s in students:
            # tuple example
            record = (s["id"], s["name"], s["city"])
            print(record)

def search_student():
    sid = input("Enter Student ID to search: ")
    found = False
    for s in students:
        if s["id"] == sid:
            print("Found:", s)
            found = True
            break
    if not found:
        print("❌ Student not found.")

def update_city():
    sid = input("Enter Student ID to update city: ")
    for s in students:
        if s["id"] == sid:
            new_city = input("Enter new city: ")
            s["city"] = new_city
            print("✅ City updated successfully!")
            return
    print("❌ Student not found.")

def delete_student():
    sid = input("Enter Student ID to delete: ")
    for s in students:
        if s["id"] == sid:
            students.remove(s)
            print("✅ Student deleted successfully!")
            return
    print("❌ Student not found.")

def show_statistics():
    print("📊 Total Students:", len(students))
    # set example
    unique_cities = set([s["city"] for s in students])
    print("Unique Cities:", unique_cities)

# Menu loop
while True:
    print("\n===== Student Record Management =====")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Search Student")
    print("4. Update City")
    print("5. Delete Student")
    print("6. Show Statistics")
    print("7. Exit")

    choice = input("Enter your choice (1-7): ")

    if choice == "1":
        add_student()
    elif choice == "2":
        view_students()
    elif choice == "3":
        search_student()
    elif choice == "4":
        update_city()
    elif choice == "5":
        delete_student()
    elif choice == "6":
        show_statistics()
    elif choice == "7":
        print("👋 Exiting program...")
        break
    else:
        print("❌ Invalid choice. Try again.")
```
